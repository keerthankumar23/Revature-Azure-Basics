```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.appName("EmployeeDF").getOrCreate()

data = [
    (101, "John", "IT", 50000, "Chennai", 30),
    (102, "Alice", "HR", 60000, "Bangalore", 28),
    (103, "Bob", "Finance", 70000, "Mumbai", 35),
    (104, "David", "IT", 80000, "Chennai", 40),
    (105, "Eva", "HR", 55000, "Hyderabad", 25),
    (106, "Frank", "Finance", 90000, "Mumbai", 45)
]

columns = ["emp_id", "emp_name", "department", "salary", "city", "age"]

emp_df = spark.createDataFrame(data, columns)
```

```python
emp_df.show()
```

```python
emp_df.printSchema()
```

```python
emp_df.select("emp_name", "salary").show()
```

```python
emp_df.select("emp_id", "emp_name", "department").show()
```

```python
emp_df.withColumnRenamed("emp_name", "employee_name").show()
```

```python
emp_df.withColumn("bonus", col("salary") * 0.10).show()
```

```python
emp_df.withColumn(
    "dept_code",
    when(col("department")=="IT","D001")
    .when(col("department")=="HR","D002")
    .otherwise("D003")
).show()
```

```python
emp_df.drop("age").show()
```

```python
# 9. Salary after 15% hike
emp_df.withColumn(
    "salary_after_hike",
    col("salary") * 1.15
).show()
```

```python
# 10. Display column names
print(emp_df.columns)
```

```python
# Section 2 – Filtering Records
# Salary > 60000
emp_df.filter(col("salary") > 60000).show()
```

```python
# IT Employees
emp_df.filter(col("department")=="IT").show()
```

```python
# Chennai Employees
emp_df.filter(col("city")=="Chennai").show()
```

```python
# Age > 30
emp_df.filter(col("age") > 30).show()
```

```python
# Age Between 25 and 35
emp_df.filter(col("age").between(25,35)).show()
```

```python
# IT and Salary > 60000
emp_df.filter(
    (col("department")=="IT") &
    (col("salary")>60000)
).show()
```

```python
# HR or Finance
emp_df.filter(
    col("department").isin("HR","Finance")
).show()
```

```python
# Salary NULL
emp_df.filter(col("salary").isNull()).show()
```

```python
# Names Start with A
emp_df.filter(col("emp_name").startswith("A")).show()
```

```python
# City Contains i
emp_df.filter(col("city").contains("i")).show()
```

```python
# Section 3 – Sorting Data
# Salary Ascending
emp_df.orderBy("salary").show()
```

```python
# Salary Descending
emp_df.orderBy(col("salary").desc()).show()
```

```python
# Department and Salary Descending
emp_df.orderBy("department", col("salary").desc()).show()
```

```python
# Top 5 Highest Paid
emp_df.orderBy(col("salary").desc()).limit(5).show()
```

```python
# Youngest Employee
emp_df.orderBy("age").show(1)
```

```python
# Section 4 – Distinct and Duplicate Handling
# Unique Departments
emp_df.select("department").distinct().show()
```

```python
# Unique Cities
emp_df.select("city").distinct().show()
```

```python
# Remove Duplicate Records
emp_df.dropDuplicates().show()
```

```python
# Remove Duplicates Based on Department
emp_df.dropDuplicates(["department"]).show()
```

```python
# Count Distinct Departments
emp_df.select("department").distinct().count()
```

```python
# Section 5 – Aggregations
# Total Salary
emp_df.agg(sum("salary")).show()
```

```python
# Average Salary
emp_df.agg(avg("salary")).show()
```

```python
# Section 6 – GroupBy Operations
# Employee Count by Department
emp_df.groupBy("department").count().show()
```

```python
# Average Salary by Department
emp_df.groupBy("department") \
      .agg(avg("salary").alias("avg_salary")) \
      .show()
```

```python
# Section 7 – Null Handling
# Records Having NULL Salary
emp_df.filter(col("salary").isNull()).show()
```

```python
# Replace NULL Salary with 50000
emp_df.na.fill({"salary":50000}).show()
```

```python
# Drop NULL Records
emp_df.na.drop().show()
```

```python
# Count NULL Salary Records
emp_df.filter(col("salary").isNull()).count()
```

```python
dept_avg = emp_df.groupBy("department") \
                 .agg(avg("salary").alias("avg_salary"))

df_join = emp_df.join(dept_avg, "department")

result = df_join.withColumn(
    "salary",
    when(col("salary").isNull(), col("avg_salary"))
    .otherwise(col("salary"))
).drop("avg_salary")

result.show()
```
