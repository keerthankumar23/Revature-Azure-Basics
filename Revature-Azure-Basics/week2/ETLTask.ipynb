```python
df = pd.read_csv("customers.csv")

df
```

```python
# Fill null values

df["Email"] = df["Email"].fillna("Not Available")

# Remove duplicates

df = df.drop_duplicates()

# Convert JoinDate

df["JoinDate"] = pd.to_datetime(df["JoinDate"])

df
```

```python
total_purchase = df["TotalPurchase"].sum()

print("Total Purchase:", total_purchase)
```

```python
city_revenue = (
    df.groupby("City")["TotalPurchase"]
      .sum()
      .reset_index()
)

city_revenue
```

```python
city_count = (
    df.groupby("City")["CustomerID"]
      .count()
      .reset_index()
)

city_count
```

```python
df.to_sql(
    name="customer_fact",
    con=engine,
    if_exists="replace",
    index=False
)

print("Customer Fact Loaded")
```

```python
city_revenue = (
    df.groupby("City")["TotalPurchase"]
      .sum()
      .reset_index()
)

city_revenue.to_sql(
    name="customer_city_summary",
    con=engine,
    if_exists="replace",
    index=False
)

print("Summary Loaded")
```

```python
import pandas as pd

pd.read_sql(
    "SELECT * FROM customer_fact",
    con=engine
)
```

```python
pd.read_sql(
    "SELECT * FROM customer_city_summary",
    con=engine
)
```

```python
pd.read_sql(
"""
SELECT COUNT(*) AS TotalCustomers
FROM customer_fact
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT CustomerName,
       TotalPurchase
FROM customer_fact
ORDER BY TotalPurchase DESC
LIMIT 5
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT City,
       SUM(TotalPurchase) AS Revenue
FROM customer_fact
GROUP BY City
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT City,
       COUNT(*) AS CustomerCount
FROM customer_fact
GROUP BY City
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT CustomerName,
       TotalPurchase,
       RANK() OVER(
       ORDER BY TotalPurchase DESC
       ) AS RankNo
FROM customer_fact
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT CustomerName,
       TotalPurchase,
       RANK() OVER(
       ORDER BY TotalPurchase DESC
       ) AS RankNo
FROM customer_fact
""",
con=engine
)
```

```python
query="""
SELECT CustomerName,
       TotalPurchase,
       RANK() OVER(
       ORDER BY TotalPurchase DESC
       ) AS RankNo
FROM customer_fact
"""
pd.read_sql(
    query,engine
)
```

```python
pd.read_sql(
"""
SELECT *
FROM customer_fact
WHERE TotalPurchase >
(
    SELECT AVG(TotalPurchase)
    FROM customer_fact
)
""",
con=engine
)
```

```python
pd.read_sql(
"""
SELECT YEAR(JoinDate) AS Year,
       MONTH(JoinDate) AS Month,
       COUNT(*) AS CustomersJoined
FROM customer_fact
GROUP BY YEAR(JoinDate),
         MONTH(JoinDate)
ORDER BY Year, Month
""",
con=engine
)
```
