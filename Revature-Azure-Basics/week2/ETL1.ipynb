```python
import pandas as pd

data = {
    "CustomerID": [101,102,103,104,105,105],
    "CustomerName": ["Rahul","Priya","Arjun","Sneha","Kiran","Kiran"],
    "City": ["Hyderabad","Chennai","Bangalore","Hyderabad","Chennai","Chennai"],
    "Email": [
        "rahul@gmail.com",
        "priya@gmail.com",
        "arjun@gmail.com",
        "sneha@gmail.com",
        None,
        None
    ],
    "Age": [25,28,30,27,29,29],
    "JoinDate": [
        "2023-01-10",
        "2023-02-15",
        "2023-03-20",
        "2023-04-25",
        "2023-05-10",
        "2023-05-10"
    ],
    "TotalPurchase": [5000,7000,9000,6000,8000,8000]
}

df = pd.DataFrame(data)

print(df)
```

```python
df.to_csv("customers.csv", index=False)
```

```python
df = pd.read_csv("customers.csv")

print(df)
```

```python
df.isnull().sum()
```

```python
df["Email"] = df["Email"].fillna("Not Available")
```

```python
df.drop_duplicates(inplace=True)
```

```python
total_purchase = df["TotalPurchase"].sum()

print("Total Purchase:", total_purchase)
```

```python
city_revenue = df.groupby("City")["TotalPurchase"].sum()

print(city_revenue)
```
