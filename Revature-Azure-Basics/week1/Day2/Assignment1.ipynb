```python
orders = [
    {"order_id": 1, "amount": 12000, "status": "Delivered"},
    {"order_id": 2, "amount": 5000, "status": "Cancelled"},
    {"order_id": 3, "amount": 15000, "status": "Delivered"},
    {"order_id": 4, "amount": 7000, "status": "Pending"}
]

premium_count = 0

for order in orders:

    if order["status"] == "Cancelled":
        continue

    if order["amount"] >= 10000:
        print(order["order_id"], "Premium Order")
        premium_count += 1
    else:
        print(order["order_id"], "Regular Order")

print("Total Premium Orders:", premium_count)
```
