```python
numbers=[45,12,78,34,78,90,12,56,23,45]
def sum_all(arr):
  t=0
  for num in arr:
    t+=num
  return t
print(sum_all(numbers))
```

```python
x= [45, 12, 78, 34, 90, 12, 56, 23, 45]
first=-1
second=-2
for i in x:
  if i>first:
    second=first
    first=i
  elif i>second and i!=first:
    second=i
print(second)
```

```python
list= [45, 12, 78, 34, 78, 90, 12, 56, 23, 45]
dup=[]
for i in range(len(list)):
  for j in range(i+1,len(list)):
    if list[i]==list[j]:
      dup.append(list[i])


print(dup)
```

```python

def student_marks(marks):
    average = sum(marks) / len(marks)
    highest = max(marks)
    lowest = min(marks)

    return average, highest, lowest


# Sample list of student marks
marks = [78, 85, 92, 67, 88]

# Function call
avg, high, low = student_marks(marks)

# Display results
print("Average Marks:", avg)
print("Highest Marks:", high)
print("Lowest Marks:", low)
```
